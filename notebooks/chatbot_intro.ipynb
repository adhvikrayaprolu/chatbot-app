```python
import openai
```

```python
api_key = "REDACTED_USE_ENVIRONMENT"
client = openai.OpenAI(api_key = api_key)
```

```python
def getLLMResponse(prompt, temperature = 0.1, model = "gpt-4.1-2025-04-14"):
    response = client.chat.completions.create(
        messages = [
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature = temperature,
        model = model
    )
    return response.choices[0].message.content

# roles include system, assistant and user
```

```python
# improvement on the previous function to include previous memories 

def getLLMResponseNew(query, temperature = 0.1, model = "gpt-4.1-2025-04-14"):
    response = client.chat.completions.create(
        messages = query,
        temperature = temperature,
        model = model
    )
    return response.choices[0].message.content
```
